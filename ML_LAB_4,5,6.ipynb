```python
import pandas as pd
import numpy as np
```

```python
# Load the Dataset

df = pd.read_csv("Automobile_data lab 4 (2).csv")
```

```python
# Verify and understand the Dataset

df.head()
```

```python
# Standardised the data in the columns

df["normalized-losses"].replace("?",np.nan,inplace=True)
```

```python
# Check for Null Values

df.isnull().sum()
```

```python
# Fill the missing Values

df["normalized-losses"] = pd.to_numeric(
    df["normalized-losses"],
    errors="coerce"
)

m=df["normalized-losses"].mean()
print("Mean of normalized-losses : ",m)
df["normalized-losses"].fillna(m,inplace=True)
```

```python
# Verify

df.isnull().sum()
```

```python
# Do the same for Horse-power Column

#replace ? with null values in "horsepower" column permanently
df["horsepower"].replace("?",np.nan,inplace=True)
#change the datatype of df["horsepower"] column from object to float
#permanently
df["horsepower"]=df["horsepower"].astype("float")

df.isnull().sum()
```

```python
m=df["horsepower"].mean()
print("Mean of horsepower : ",m)
df["horsepower"].fillna(m,inplace=True)
```

```python
# separate numerical type column and object type column and store in new dataset

df_num=df.select_dtypes(["int64","float64"])#hold to int and float type data
df_cat=df.select_dtypes(object)#hold object type data
```

```python
# Encode the Categorical Columns

from sklearn.preprocessing import LabelEncoder
for col in df_cat :
    le=LabelEncoder() #create the object of LabelEncoder class
    df_cat[col]=le.fit_transform(df_cat[col])
```

```python
# Verify

df_cat.head()
```

```python
# Concatenation of both dataset df_num and df_cat and hold new dataset df_new

df_new=pd.concat([df_num,df_cat],axis=1) #axis=1 column wise
df_new.head()
```

```python
# Select independent means input variables(X) and target means dependent variable(Y)

X=df_new.drop("price",axis=1)
#apart from price, all columns hold in X variable
Y=df_new["price"]
#output variable
```

```python
# Train test split 70%-30%

from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,
                                              random_state=1)
```

```python
# Check the shape of test and train data

X_train.shape,X_test.shape
```

```python
# Apply StandardScaler on X_train and X_test

from sklearn.preprocessing import StandardScaler
ss=StandardScaler()
X_train=ss.fit_transform(X_train)
X_test=ss.transform(X_test)
```

```python
# Verify

X_train
```

```python
# Task 2: Build and Train Linear Regression Model
# 1 Import Linear Regression

from sklearn.linear_model import LinearRegression
```

```python
# 2 Create Model Object

lr=LinearRegression()
```

```python
# 3 Train the Model with 70% data

lr.fit(X_train,Y_train)
```

```python
# Task 3: Predict Vehicle Prices

# Generate Predictions
y_pred = lr.predict(X_test)

```

```python
# View Predicted Prices

y_pred[:10]
```

```python
# ML LAB 5 - Evaluate Regression Model Performance
# Check the training score on training data

lr.score(X_train,Y_train)
```

```python
# Check slope or coefficient of each input , use inbuilt method coef_

lr.coef_
```

```python
# Check testing score with 30% data

lr.score(X_test,Y_test)
```

```python
# Create a dictionary which hold input column and coefficient of input

dict={'features':X.columns,'Slope':lr.coef_}
df1=pd.DataFrame(dict)
df1
```

```python
# Task 2: Apply Lasso Regression
# 1 Call inbuilt class Lasso

from sklearn.linear_model import Lasso
```

```python
# 2 Select the right value of lambda /alpha in lasso, use simple technique

for i in range(200,1001,50): #start=200 stop=1001-1=1000 step=+50
    #create the object of Lasso class , passing the value of lambda
    l1=Lasso(i) #first i=200 , next i=250 ......
    #automatic write formula RSS+i*sum(abs(all coef))
    #train the model with 70% data using Lasso class , use fit() inbuilt
    #method of Lasso class
    l1.fit(X_train,Y_train)
    print("Value of Lambda : ",i)
    #check training score
    print("Training Score : ",l1.score(X_train,Y_train))
    #check testing score
    print("Testing Score : ",l1.score(X_test,Y_test))
```

```python
# 3 Check coefficient /slope of each input/features after apply lasso

l1.coef_
```

```python
# 4 Add new column of slope in existing dataframe df1 after apply Lasso

df1['Slope_Lasso']=l1.coef_
df1
```

```python
# 5 Create the Lasso Regression Model

lasso_model = Lasso(alpha= 200)
```

```python
# 6 Train the Model

lasso_model.fit(X_train, Y_train)
```

```python
#  7 Generate Predictions

lasso_pred = lasso_model.predict(X_test)
```

```python
# 8 Create the Lasso Regression Model

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
print("MAE :", mean_absolute_error(Y_test, lasso_pred))
print("MSE :", mean_squared_error(Y_test, lasso_pred))
print("R² Score :", r2_score(Y_test, lasso_pred))
```

```python
# Task 3: Apply Ridge Regression
# Call Ridge class inbuilt class

from sklearn.linear_model import Ridge
```

```python
# 2 Find range of lambda 0-50

for i in range(0,51): #start=0 stop=51-1=50 step=+1
    #create the object of Ridge means L2 and passing the value of lambda
    l2=Ridge(i)
    #train the model with 70% data using Ridge class , use fit() inbuilt
    #method of Ridge class
    l2.fit(X_train,Y_train)
    print("Value of Lambda : ",i)
    #check training score
    print("Training Score : ",l2.score(X_train,Y_train))
    #check testing score
    print("Testing Score : ",l2.score(X_test,Y_test))
```

```python
# 3 Check coefficient after apply Ridge

l2.coef_
df1['Slope_Ridge']=l2.coef_
df1
```

```python
# 4 Create the Ridge Regression Model

ridge_model = Ridge(alpha= 12)
```

```python
# 5 Train the Model

ridge_model.fit(X_train, Y_train)
```

```python
# 6 Generate Predictions

ridge_pred = ridge_model.predict(X_test)
```

```python
# 7 Evaluate the Ridge Model

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
print("MAE :", mean_absolute_error(Y_test, ridge_pred))
print("MSE :", mean_squared_error(Y_test, ridge_pred))
print("R² Score :", r2_score(Y_test, ridge_pred))
```
